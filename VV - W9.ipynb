```python
24.

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import load_sample_image
from sklearn.cluster import KMeans
from sklearn.utils import shuffle

china = load_sample_image("china.jpg")

china_normalized = np.array(china, dtype=np.float64) / 255
height, width, channels = china_normalized.shape

pixels = china_normalized.reshape(-1, 3)

n_colors = 16
pixels_sample = shuffle(pixels, random_state=42, n_samples=1000)

kmeans = KMeans(n_clusters=n_colors, random_state=42, n_init=10)
kmeans.fit(pixels_sample)

labels = kmeans.predict(pixels)
compressed_image = kmeans.cluster_centers_[labels].reshape(height, width, channels)

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.imshow(china)
plt.axis("off")
plt.title("Original Image")

plt.subplot(1, 2, 2)
plt.imshow(compressed_image)
plt.axis("off")
plt.title(f"Compressed Image using K-Means with {n_colors} colors")

plt.show()
```
